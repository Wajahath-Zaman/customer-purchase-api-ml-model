```python
import joblib
import pandas as pd
```

```python
model = joblib.load(
    "customer_purchase_model.pkl"
)
```

```python
new_customer = pd.DataFrame({
    "age": [32],
    "income": [75000],
    "previous_purchase": [5],
    "city": ["Hyderabad"],
    "device": ["Mobile"]
})
```

```python
prediction = model.predict(new_customer)
probability = model.predict_proba(new_customer)[:,1]
```

```python
print(prediction, probability)
```
